```python
mport numpy as np
import nibabel as nib
import trimesh
from scipy.ndimage import binary_fill_holes

# Load the STL file
mesh = trimesh.load_mesh(r"C:\Users\ferat\Desktop\suzanne.stl")

# Define voxel grid resolution
resolution = 128  # Adjust for finer details

# Get the bounding box and create a voxel grid
grid = mesh.voxelized(pitch=mesh.extents.max() / resolution).fill()
voxel_data = grid.matrix.astype(np.uint8)  # Convert to binary (1 for filled, 0 for empty)

# Fill holes to simulate a solid MRI-style structure
filled_data = binary_fill_holes(voxel_data).astype(np.uint8)

# Create an affine transformation matrix (identity for now)
affine = np.eye(4)

# Save as NIfTI
nii_img = nib.Nifti1Image(filled_data, affine)
nib.save(nii_img, "susan_mri.nii.gz")

print("Saved NIfTI image as 'susan_mri.nii.gz'")
```
